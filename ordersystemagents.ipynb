```python
import pyodbc
from ollama import chat

# ============================================================
# 1. SQL SERVER CONNECTION
# ============================================================

connection = pyodbc.connect(
    "DRIVER={ODBC Driver 18 for SQL Server};"
    "SERVER=DESKTOP-NSQVUD1;"
    "DATABASE=orderssystem;"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)

cursor = connection.cursor()


# ============================================================
# 2. SALES ANALYSIS AGENT
# ============================================================

def sales_analysis_agent(user_prompt):

    # Get required sales data
    cursor.execute("""
        SELECT
            o.orderid,
            c.custid,
            c.custname,
            s.salesmanid,
            s.salesmanname,
            o.amount,
            o.dateofpurchase
        FROM orders o
        INNER JOIN customer c
            ON c.custid = o.custid
        INNER JOIN salesman s
            ON s.salesmanid = o.salesmanid
        ORDER BY o.amount DESC
    """)

    rows = cursor.fetchall()

    # --------------------------------------------------------
    # Convert database records into text for Ollama
    # --------------------------------------------------------

    sales_data = "\n".join(
        f"Order ID: {r.orderid}, "
        f"Customer ID: {r.custid}, "
        f"Customer: {r.custname}, "
        f"Salesman ID: {r.salesmanid}, "
        f"Salesman: {r.salesmanname}, "
        f"Amount: {r.amount}, "
        f"Date: {r.dateofpurchase}"
        for r in rows
    )

    # --------------------------------------------------------
    # Send data + question to Ollama
    # --------------------------------------------------------

    response = chat(
        model="llama3.2",
        messages=[
            {
                "role": "system",
                "content": """
You are a Sales Analysis Agent.

Your responsibilities are:

1. Analyze customer sales.
2. Analyze salesman performance.
3. Analyze orders.
4. Calculate total sales.
5. Calculate average sales.
6. Find highest and lowest sales.
7. Compare sales between years.
8. Identify top customers.
9. Identify top salesmen.
10. Answer questions only from the supplied database data.

Do not invent database records.

If a calculation is required, calculate it from the supplied data.
"""
            },
            {
                "role": "user",
                "content": f"""
SALES DATABASE:

{sales_data}

USER QUESTION:

{user_prompt}

Provide a clear and accurate answer.
"""
            }
        ]
    )

    return response.message.content


# ============================================================
# 3. USER INTERFACE
# ============================================================

print("\n======================================")
print("       SALES ANALYSIS AGENT")
print("======================================")

while True:

    user_prompt = input(
        "\nAsk your sales question "
        "(type 'exit' to quit): "
    )

    if user_prompt.lower() == "exit":
        break

    result = sales_analysis_agent(user_prompt)

    print("\nAI SALES ANALYSIS:")
    print("--------------------------------------")
    print(result)


# ============================================================
# 4. CLOSE CONNECTION
# ============================================================

cursor.close()
connection.close()
```
