```python

import pyodbc
from ollama import chat

# ============================================================
# 1. SQL SERVER CONNECTION
# ============================================================

connection = pyodbc.connect(
    "DRIVER={ODBC Driver 18 for SQL Server};"
    "SERVER=DESKTOP-NSQVUD1;"
    "DATABASE=orderssystem;"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)

cursor = connection.cursor()


# ============================================================
# 2. CUSTOMER PERFORMANCE AGENT
# ============================================================

def customer_performance_agent(user_prompt):

    # --------------------------------------------------------
    # Get customer + order information
    # --------------------------------------------------------

    cursor.execute("""
        SELECT
            c.custid,
            c.custname,
            c.custaddress,
            o.orderid,
            o.amount,
            o.dateofpurchase,
            s.salesmanid,
            s.salesmanname
        FROM customer c
        LEFT JOIN orders o
            ON c.custid = o.custid
        LEFT JOIN salesman s
            ON s.salesmanid = o.salesmanid
        ORDER BY c.custid, o.dateofpurchase
    """)

    rows = cursor.fetchall()


    # --------------------------------------------------------
    # Convert database records into text
    # --------------------------------------------------------

    customer_data = "\n".join(
        f"Customer ID: {r.custid}, "
        f"Customer: {r.custname}, "
        f"Address: {r.custaddress}, "
        f"Order ID: {r.orderid}, "
        f"Amount: {r.amount}, "
        f"Date: {r.dateofpurchase}, "
        f"Salesman ID: {r.salesmanid}, "
        f"Salesman: {r.salesmanname}"
        for r in rows
    )


    # --------------------------------------------------------
    # Send customer data + question to Ollama
    # --------------------------------------------------------

    response = chat(
        model="llama3.2",
        messages=[
            {
                "role": "system",
                "content": """
You are a Customer Performance Agent.

Your responsibilities are:

1. Analyze customer performance.
2. Identify top customers.
3. Calculate total purchases by customer.
4. Calculate number of orders by customer.
5. Calculate average order value.
6. Find the highest-value customer.
7. Find the lowest-value customer.
8. Find customers with the highest number of orders.
9. Compare customers.
10. Identify customers with high purchase values.
11. Identify customers with low purchase values.
12. Analyze customer purchase history.
13. Analyze customer relationships with salesmen.
14. Answer questions using ONLY the supplied database data.

Important rules:

- Do not invent customers.
- Do not invent orders.
- Do not invent sales amounts.
- Do not invent dates.
- Do not assume information that is not present.
- Perform calculations using the supplied data.
- If the requested customer does not exist, clearly say so.
- Give clear and simple business explanations.
"""
            },
            {
                "role": "user",
                "content": f"""
CUSTOMER DATABASE:

{customer_data}

USER QUESTION:

{user_prompt}

Analyze the database and provide a clear and accurate answer.
"""
            }
        ]
    )

    return response.message.content


# ============================================================
# 3. USER INTERFACE
# ============================================================

print("\n======================================")
print("      CUSTOMER PERFORMANCE AGENT")
print("======================================")

while True:

    user_prompt = input(
        "\nAsk your customer question "
        "(type 'exit' to quit): "
    )

    if user_prompt.lower() == "exit":
        break

    result = customer_performance_agent(user_prompt)

    print("\nAI CUSTOMER ANALYSIS:")
    print("--------------------------------------")
    print(result)


# ============================================================
# 4. CLOSE CONNECTION
# ============================================================

cursor.close()
connection.close()

```
